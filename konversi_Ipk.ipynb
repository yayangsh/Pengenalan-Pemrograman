```python
#tugas: memproses transkip nilai akademik

#Gunakan metode split(".") untuk memisahkan teks input menjadi variabel
#nama,nim, nilai_tugas, nilai_uts, dan nilai_uas.

# 1. Input data mahasiswa
data = input("Masukkan data mahasiswa: ")

# 2. Memisahkan data menggunakan split()
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

# 3. Menghilangkan spasi yang tidak diperlukan
nama = nama.strip()
nim = nim.strip()

# 4.Konversi ketiga nilai ke bentuk float.

nilai_tugas = float(nilai_tugas)
nilai_uts = float(nilai_uts)
nilai_uas = float(nilai_uas)

# 5.Hitung nilai akhir dengan bobot:

bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45
nilai_akhir = nilai_tugas * 0.20 + nilai_uts * 0.35 + nilai_uas * 0.45

# 6.Tampilkan tabel ringkasan

print(f"┌{"─"*41}┐")
print(f"│{'TRANSKIP NILAI AKADEMIK':^41}│")
print(f"├{"─"*41}┤")
print(f"│ Nama : {nama:<33}│")
print(f"│ NIM  : {nim:<33}│")
print(f"├{"─" * 21}┬{"─" * 9}┬{"─" * 9}┤")
print(f"│ {'Komponen':<19} │ {'Nilai':<7} │ {'Bobot':<7} │")
print(f"├{"─" * 21}┼{"─" * 9}┼{"─" * 9}┤")
print(f"│ {'Tugas':<19} │ {nilai_tugas:<7.1f} │ {bobot_tugas:<7.0%} │")
print(f"│ {'UTS':<19} │ {nilai_uts:<7.1f} │ {bobot_uts:<7.0%} │")
print(f"│ {'UAS':<19} │ {nilai_uas:<7.1f} │ {bobot_uas:<7.0%} │")
print(f"├{"─" * 21}┴{"─" * 9}┴{"─" * 9}┤")
print(f"│ NILAI AKHIR         : {nilai_akhir:<17.1f} │")
print(f"└{"─"*41}┘")
```
